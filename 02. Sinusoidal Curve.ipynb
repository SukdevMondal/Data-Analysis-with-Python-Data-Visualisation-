```python
#By using linspace function

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 8*np.pi, 900)    # Taking 8 cycle. Each cycle contains 900 x values
y = np.sin(x)
plt.plot(x, y, color='green', linewidth=2)
plt.axhline(y=0, color='red', linestyle='--', linewidth=1)
plt.xlabel('X-axis', color='blue')
plt.ylabel('Y-axis', color='blue')
plt.title('Sine Wave', color='magenta')
```

```python
#By using arange function

import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 8*np.pi, 0.05)    # Taking 8 cycle. Each cycle contains 900 x values
y = np.sin(x)
plt.plot(x, y, color='green', linewidth=2)
plt.axhline(y=0, color='red', linestyle='--', linewidth=1)
plt.xlabel('X-axis', color='blue')
plt.ylabel('Y-axis', color='blue')
plt.title('Sine Wave', color='magenta')
```
