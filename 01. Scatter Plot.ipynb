```python
import matplotlib.pyplot as plt
x = [2, 5, 3, 6, 4, 9]
y = [5, 7, 3, 2, 6, 7]
plt.scatter(x, y)
plt.xlabel('Age (Years)')
plt.ylabel('Score (Out of 10)')
plt.title('Score of different age')
```
