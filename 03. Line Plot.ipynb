```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 6*np.pi, 100)    # Generating the x valuee
ys = np.sin(x)     # Generating the y values
yc = np.cos(x)
plt.figure(figsize=(10, 5))     # Defining size of the figure
plt.plot(x, ys, color='Green', label='Sin(x)', linestyle='-', linewidth=2)
plt.scatter(x, ys, color='red', marker= 's')
plt.scatter(x, yc, color='red', marker= 'o')
plt.plot(x, yc, color='Blue', label='Cos(x)', linestyle='--', linewidth=2)
plt.legend(loc='upper right', shadow=True)
plt.axhline(color='black', linewidth=1)
plt.grid(True, linestyle='--', alpha=0.5)     # alpha value is for visibility (0 to 1)
plt.xlabel('X-axis', color='brown')
plt.ylabel('Y-axis',  color='brown')
plt.title('Sine wave and Cos wave', color='magenta')
```
