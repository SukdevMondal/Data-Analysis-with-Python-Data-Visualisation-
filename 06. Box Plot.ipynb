```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

data = [3, 4, 6, 7, 2, -7, 3, 8, 12, 11, 9, 5, 18, 1, 7]
df1 = pd.DataFrame(data)
q1 = np.percentile(df1, 25)
q3 = np.percentile(df1, 75)
iqr = q3-q1
upw = q3 + (1.5*iqr)
low = q1 - (1.5*iqr)

plt.boxplot(df1)
plt.title('Box Plot', color='red')
plt.xlabel('Number', color='Green')
plt.ylabel('Frequency', color='Blue')
```
