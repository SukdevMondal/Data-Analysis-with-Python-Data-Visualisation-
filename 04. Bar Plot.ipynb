```python
import matplotlib.pyplot as plt
student=['Sukdev', 'Anukul', 'Akhi', 'Alok', 'Rakhi', 'Suvro', 'Tarok']
score=[98, 93, 83, 63, 74, 72, 56]
plt.bar(student, score, color='green')
plt.xlabel('Name of students', color='blue')
plt.ylabel('Score', color='red')
plt.title('Students score in mathematics')
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = pd.read_excel("D:\Documents\Research\Projects\Cold Storage\Python practice.xlsx")
df1=pd.DataFrame(data)
df2=data[['Attributes', 'Average']].head(5)    # Creating a new dataframe taking two columns and first rows
df3=df2.fillna(df2.mean(numeric_only=True))
plt.figure(figsize=(10,5))    # Figure size should be defined before plot
bar=plt.bar(df3.Attributes, df3.Average, color='orange')
plt.bar_label(bar)     # Showing the values of each bar above the bar
plt.xlabel('Attributes')
plt.ylabel('Rating')
plt.title('Fruit attributes and rating', color='blue')
plt.show()
```
