```python
# By Matplotlib

import matplotlib.pyplot as plt
Result= ['CGPA: 4.00-3.50', 'CGPA: 3.50-3.00', 'CGPA: 3.00-2.50', 'CGPA: 2.50-2.00', 'CGPA: <2.00']
Students=[6, 12, 27, 19, 8]
plt.pie(Students, labels= Result, autopct='%1.2f%%')
plt.title('Result wise students percentage', color='red', loc='center')
```

```python
# By Plotly

import plotly.graph_objects as go
Result= ['CGPA: 4.00-3.50', 'CGPA: 3.50-3.00', 'CGPA: 3.00-2.50', 'CGPA: 2.50-2.00', 'CGPA: <2.00']
Students=[6, 12, 27, 19, 8]
fig=go.Figure(data=[go.Pie(labels=Result, values=Students, hole=0.5)])
fig.show()
```
